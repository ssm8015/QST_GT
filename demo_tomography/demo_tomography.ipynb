```python
from tomography_original_functions import (
    available_methods,
    run_paper_demo,
    print_demo_summary,
)

print("Qubit methods:", available_methods("qubit"))
print("Qudit / qutrit / hybrid methods:", available_methods("hybrid"))
```

## Demo controls


Use:
- `SYSTEM = "qubit"` for Pauli-string tomography design with sensing matrix `A`.
- `SYSTEM = "qutrit"` for homogeneous qutrit operator partitioning.
- `SYSTEM = "hybrid"` for mixed qubit-qutrit operator partitioning.

For qubit systems, `OPERATOR_COUNT = None` uses all non-identity Pauli operators.

```python
# ---------------- DEMO CONFIGURATION ----------------
SYSTEM = "qubit"      # "qubit", "qutrit", "qudit", or "hybrid"
METHOD = "dsatur"     # qubit: "dsatur", "rlf", "spectral", "nn", "ilp", "combined", "all"
                      # qutrit/hybrid: "dsatur", "rlf", "spectral", "gnn", "ilp", "all"

# System size
N_QUBITS = 4           # used for qubit and hybrid
N_QUTRITS = 1          # used for qutrit and hybrid
LOCAL_DIM = 3          # used only when SYSTEM = "qudit"
N_QUDITS = 1           # used only when SYSTEM = "qudit"

# Operator set selection
OPERATOR_COUNT = None  # None = all for qubit; default small value for qutrit/hybrid
OPERATOR_SELECTION = "all"  # qubit only: "all", "random", or "first"
TARGET_OPERATOR_LABELS = None
# Example for a fixed two-qubit target set:
# TARGET_OPERATOR_LABELS = ["IX", "XI", "XX", "IY", "YI", "YY", "IZ", "ZI", "ZZ"]

# Numerical/demo controls
SEED = 250
DSATUR_ATTEMPTS = 25
GNN_EPOCHS = 150
VERBOSE = True
```

## Run the pipeline

```python
result = run_paper_demo(
    system=SYSTEM,
    method=METHOD,
    operator_count=OPERATOR_COUNT,
    operator_selection=OPERATOR_SELECTION,
    target_operator_labels=TARGET_OPERATOR_LABELS,
    n_qubits=N_QUBITS,
    n_qutrits=N_QUTRITS,
    local_dim=LOCAL_DIM,
    n_qudits=N_QUDITS,
    seed=SEED,
    dsatur_attempts=DSATUR_ATTEMPTS,
    gnn_epochs=GNN_EPOCHS,
    verbose=VERBOSE,
)
```

```python
print_demo_summary(result)
```
